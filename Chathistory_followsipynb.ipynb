```python
from google.colab import files

uploaded = files.upload()
pdf_path = next(iter(uploaded))
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

pages = PyPDFLoader(pdf_path).load()

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(pages)

print("Pages:", len(pages))
print("Chunks:", len(chunks))
```

```python
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

db = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model,
    collection_name="lesson16_pdf"
)

print("ChromaDB created successfully!")
```

```python
retriever = db.as_retriever(
    search_kwargs={"k": 3}
)

print("Retriever created successfully!")
```

```python
question = input("Ask a question about your PDF: ")

docs = retriever.invoke(question)

for doc in docs:
    print("\nText:", doc.page_content)
    print("Page:", doc.metadata.get("page", 0) + 1)
```

```python
def save_chat(question, answer):
    chat_history.append({
        "question": question,
        "answer": answer
    })
```

```python
save_chat(
    "What is sql?",
    "sql allows computers to learn store from data."
)

save_chat(
    "Give one example.",
    "An email spam detector is one example."
)
for message in chat_history:
    print("User:", message["question"])
    print("Assistant:", message["answer"])
    print()
print(chat_history)
```
